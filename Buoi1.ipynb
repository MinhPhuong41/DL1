```python
#3
# --- TRƯỜNG HỢP 1: MẢNG 1 CHIỀU ---
print("--- SO SÁNH MẢNG 1 CHIỀU ---")
a1 = np.array([10, 20, 30, 40])
b1 = np.array([15, 20, 25, 45])
print("Mảng a1: {a1}")
print("Mảng b1: {b1}")
print("a1 > b1  (greater):      ", np.greater(a1, b1))
print("a1 >= b1 (greater_equal):", np.greater_equal(a1, b1))
print("a1 < b1  (less):         ", np.less(a1, b1))
print("a1 <= b1 (less_equal):   ", np.less_equal(a1, b1))
# --- TRƯỜNG HỢP 2: MẢNG 2 CHIỀU (3x3) ---
print("\n--- SO SÁNH MẢNG 2 CHIỀU (3x3) ---")
# Tạo 2 mảng ngẫu nhiên để so sánh
a2 = np.random.randint(0, 10, size=(3, 3))
b2 = np.random.randint(0, 10, size=(3, 3))
print("Mảng a2:\n", a2)
print("Mảng b2:\n", b2)

print("\nKết quả greater (a2 > b2):")
print(np.greater(a2, b2))

print("\nKết quả greater_equal (a2 >= b2):")
print(np.greater_equal(a2, b2))

print("\nKết quả less (a2 < b2):")
print(np.less(a2, b2))

print("\nKết quả less_equal (a2 <= b2):")
print(np.less_equal(a2, b2))


```
