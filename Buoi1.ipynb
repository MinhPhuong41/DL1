```python
#3
# --- TRƯỜNG HỢP 1: MẢNG 1 CHIỀU ---
print("--- SO SÁNH MẢNG 1 CHIỀU ---")
a1 = np.array([10, 20, 30, 40])
b1 = np.array([15, 20, 25, 45])
print("Mảng a1: {a1}")
print("Mảng b1: {b1}")
print("a1 > b1  (greater):      ", np.greater(a1, b1))
print("a1 >= b1 (greater_equal):", np.greater_equal(a1, b1))
print("a1 < b1  (less):         ", np.less(a1, b1))
print("a1 <= b1 (less_equal):   ", np.less_equal(a1, b1))
# --- TRƯỜNG HỢP 2: MẢNG 2 CHIỀU (3x3) ---
print("\n--- SO SÁNH MẢNG 2 CHIỀU (3x3) ---")
# Tạo 2 mảng ngẫu nhiên để so sánh
a2 = np.random.randint(0, 10, size=(3, 3))
b2 = np.random.randint(0, 10, size=(3, 3))
print("Mảng a2:\n", a2)
print("Mảng b2:\n", b2)

print("\nKết quả greater (a2 > b2):")
print(np.greater(a2, b2))

print("\nKết quả greater_equal (a2 >= b2):")
print(np.greater_equal(a2, b2))

print("\nKết quả less (a2 < b2):")
print(np.less(a2, b2))

print("\nKết quả less_equal (a2 <= b2):")
print(np.less_equal(a2, b2))
```

```python
#10
n = 10
matrix = []

for i in range(n):
    row = []
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            row.append(1)
        else:
            row.append(0)
    matrix.append(row)

print("Ma trận 10x10:")
for row in matrix:
    print(*row)
```

```python
#9 Hãy tạo ngẫu nhiên tensor kích thước (10, 10, 3). Cắt thành tensor có kích thước (5, 5, 3) (sử dụng random_crop)
X = np.random.rand(10, 10, 3)
crop_h, crop_w = 5, 5

start_h = np.random.randint(0, 10 - crop_h + 1)
start_w = np.random.randint(0, 10 - crop_w + 1)

cropped = X[start_h:start_h+crop_h, start_w:start_w+crop_w, :]

print("Shape ban đầu:", X.shape)
print("Shape sau crop:", cropped.shape)
```

```python
#10 Tạo ngẫu nhiên hai tensor 1-chiều x, y có 5 phần tử. Trả về x + y nếu x < y, và x – y nếu x > y
x = np.random.randint(0, 10, 5)
y = np.random.randint(0, 10, 5)

print("x:", x)
print("y:", y)

result = np.where(x < y, x + y, x - y)

print("Kết quả:", result)
```
